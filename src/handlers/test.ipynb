```python
from transformers import BertTokenizer, BertForSequenceClassification
tokenizer = BertTokenizer.from_pretrained("../../model")
model = BertForSequenceClassification.from_pretrained("../../model")
```

```python
import torch
import torch.nn.functional as F

inputs = tokenizer("what the fuck is that", return_tensors="pt", truncation=True, padding=True)
with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits.squeeze().cpu().numpy()
    print("Logits:", logits)
    probs = F.softmax(outputs.logits, dim=1).squeeze().cpu().numpy()
    print("Probs:", probs)
```

```python
inputs = tokenizer("how are you", return_tensors="pt", truncation=True, padding=True)
with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits.squeeze()
    print("Logits:", logits)
    probs_sigmoid = torch.sigmoid(logits)
    probs_softmax = torch.softmax(logits, dim=0)
    print("Sigmoid probs:", probs_sigmoid)
    print("Softmax probs:", probs_softmax)
```
